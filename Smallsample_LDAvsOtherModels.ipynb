```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns
import datetime

from sklearn.discriminant_analysis import LinearDiscriminantAnalysis, QuadraticDiscriminantAnalysis
from sklearn.metrics import confusion_matrix, classification_report
```

```python
train_file = './archive/model_2016_ST4000DM000.csv'
test_file = './archive/model_2017_ST4000DM000.csv'
```

```python
df = pd.read_csv(train_file)
df_t = pd.read_csv(test_file)
```

```python
df
```

```python
df.info()
```

```python
df = df.dropna(axis=1, how='all')
df = df.dropna(axis=0, how='any')
df
```

```python
df = df.loc[:, df.apply(lambda col: col.max() != col.min())]
df
```

```python
df_failed = df[df['failure']==1]
df_normal = df[df['failure']==0]
print(len(df_failed))
print(len(df_normal))
```

```python
df_t = df_t.dropna(axis=1, how='all')
df_t = df_t.dropna(axis=0, how='any')
df_t = df_t.loc[:, df_t.apply(lambda col: col.max() != col.min())]
df_t
```

```python
df_train_target = df['failure']
df_test_target = df_t['failure']
df_train = df.drop(columns=['date','serial_number','failure'])
df_test = df_t.drop(columns=['date','serial_number','failure'])
```

```python
lda = LinearDiscriminantAnalysis()
lda.fit(df_train,df_train_target)
y_pred = lda.predict(df_test)
print(cofusion_matrix(df_test_target, y_pred))
print(classification_report(df_test_target, y_pred, target_names=['normal','failed']))
```
