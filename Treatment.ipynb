```python
import numpy as np
from sklearn import linear_model
from matplotlib import pyplot as plt
```

```python
# 擬似乱数生成器のシードを設定
np.random.seed(2022)
```

```python
#一つのデータセットを生成する
def treatment_parent_data_generate(alpha = 1,
                     sigma   = 1,
                     n       = 20):
    ############## パラメータ ###############
    # X:介入、Y:アウトカム、Z:介入の親
    # M = epsilon_3
    # X = alpha_2 * Z + c_2 + epsilon_2
    # Y = alpha_1 * X + c_1 + epsilon_1

    M_1 = sigma * np.random.standard_normal(n)
    M_2 = sigma * np.random.standard_normal(n)
    X = M_1 + M_2 + sigma * np.random.standard_normal(n)
    Y = M_1 + alpha * X + sigma * np.random.standard_normal(n)
    M_3 = X + Y + sigma * np.random.standard_normal(n)

    return [X, M_1, M_2, M_3, Y]
```

```python
def experiment_treatment_parent(alpha = 1,
                     sigma   = 1,
                     n       = 20,
                     M       = 1000):
    result = np.zeros([8,M])
    for i in range(M):
        data = treatment_parent_data_generate(alpha,sigma,n)
        data = np.array(data) #[X, M_1, M_2, M_3, Y]
        # [X]
        model = linear_model.LinearRegression()
        model.fit(X = data[0,:].reshape(-1,1),y = data[4,:])
        result[0,i] = model.coef_[0]
        # [X,M_1]
        model = linear_model.LinearRegression()
        model.fit(X = np.transpose(data[[0,1],:]),y = data[4,:])
        result[1,i] = model.coef_[0]
        # [X,M_2]
        model = linear_model.LinearRegression()
        model.fit(X = np.transpose(data[[0,2],:]),y = data[4,:])
        result[2,i] = model.coef_[0]
        # [X,M_3]
        model = linear_model.LinearRegression()
        model.fit(X = np.transpose(data[[0,3],:]),y = data[4,:])
        result[3,i] = model.coef_[0]
        # [X,M_1,M_2]
        model = linear_model.LinearRegression()
        model.fit(X = np.transpose(data[[0,1,2],:]),y = data[4,:])
        result[4,i] = model.coef_[0]
        # [X,M_1,M_3]
        model = linear_model.LinearRegression()
        model.fit(X = np.transpose(data[[0,1,3],:]),y = data[4,:])
        result[5,i] = model.coef_[0]
        # [X,M_2,M_3]
        model = linear_model.LinearRegression()
        model.fit(X = np.transpose(data[[0,2,3],:]),y = data[4,:])
        result[6,i] = model.coef_[0]
        # [X,M_1,M_2,M_3]
        model = linear_model.LinearRegression()
        model.fit(X = np.transpose(data[[0,1,2,3],:]),y = data[4,:])
        result[7,i] = model.coef_[0]
    return result
```

```python
result = np.array(experiment_treatment_parent(alpha = 1,
                     sigma   = 1,
                     n       = 20,
                     M       = 1000))
```
