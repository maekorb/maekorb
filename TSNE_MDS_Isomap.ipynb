```python
from sklearn.datasets import load_breast_cancer
breast_cancer = load_breast_cancer()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
%matplotlib inline
```

```python
breast_cancer_X = breast_cancer.data
y = breast_cancer.target
```

```python
from sklearn.decomposition import PCA
pca = PCA()
breast_cancer_pca = pca.fit_transform(breast_cancer_X)
breast_cancer_pca[:30]
```

```python
pca.explained_variance_ratio_
```

```python
from sklearn import preprocessing

breast_cancer_X_scaled = preprocessing.scale(breast_cancer_X)
pca = PCA(n_components=2)
breast_cancer_X_scaled = pca.fit_transform(breast_cancer_X_scaled)
```

```python
from sklearn.manifold import TSNE, MDS, Isomap

breast_cancer_tsne = TSNE(learning_rate=200,random_state=0).fit_transform(breast_cancer_X)
breast_cancer_MDS = MDS(n_components = 2).fit_transform(breast_cancer_X)
breast_cancer_ISO = Isomap(n_components = 2).fit_transform(breast_cancer_X)


plt.figure(figsize=(12, 8))
plt.subplot(221)
plt.title('PCA')
plt.scatter(breast_cancer_pca [:, 0], breast_cancer_pca [:, 1], c=y)

plt.subplot(222)
plt.scatter(breast_cancer_tsne[:, 0], breast_cancer_tsne[:, 1], c=y)
plt.title('TSNE')

plt.subplot(223)
plt.scatter(breast_cancer_MDS[:, 0], breast_cancer_MDS[:, 1], c=y)
plt.title('MDS')

plt.subplot(224)
plt.scatter(breast_cancer_ISO[:, 0], breast_cancer_ISO[:, 1], c=y)
plt.title('ISO')
```
